```python
import time
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import stumpy
from sklearn.ensemble import IsolationForest
from sklearn.metrics import f1_score, precision_score, recall_score, roc_auc_score
from sklearn.preprocessing import StandardScaler
from statsmodels.tsa.seasonal import STL

warnings.filterwarnings("ignore")

# ==== CONFIG ====
CONSOLIDATED = Path("/content/consolidated")   # change if needed
OUT_DIR = Path("/content/outputs/benchmark")

TOP5 = [1319, 1258, 439, 247, 1225]
LEAD_RANK = {1319: 1, 1258: 2, 439: 3, 247: 4, 1225: 5}

OUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
def _gt(df):
    return np.where(df["anomaly"].isna(), 0, df["anomaly"].fillna(0).astype(int)).astype(int)

def _add_calendar(df):
    df = df.copy()
    df["hour"] = df["timestamp"].dt.hour
    df["dayofweek"] = df["timestamp"].dt.dayofweek
    df["month"] = df["timestamp"].dt.month
    return df

def metrics(pred_binary, pred_scores, gt_binary):
    if gt_binary.sum() == 0:
        return dict(precision=np.nan, recall=np.nan, f1=np.nan, auroc=np.nan,
                    n_detected=0, n_tp=0, n_fp=0, n_fn=0)

    prec = precision_score(gt_binary, pred_binary, zero_division=0)
    rec = recall_score(gt_binary, pred_binary, zero_division=0)
    f1 = f1_score(gt_binary, pred_binary, zero_division=0)

    try:
        s_mn, s_mx = pred_scores.min(), pred_scores.max()
        sn = (pred_scores - s_mn) / (s_mx - s_mn) if s_mx > s_mn else pred_scores
        auroc = roc_auc_score(gt_binary, sn)
    except:
        auroc = np.nan

    return dict(
        precision=float(prec), recall=float(rec), f1=float(f1), auroc=float(auroc),
        n_detected=int(pred_binary.sum()),
        n_tp=int((pred_binary & gt_binary).sum()),
        n_fp=int((pred_binary & (1 - gt_binary)).sum()),
        n_fn=int(((1 - pred_binary) & gt_binary).sum()),
    )
```

```python
def stl_residual(y, period=24):
    if len(y) < 2 * period + 1:
        return y.copy()
    try:
        return STL(y, period=period, robust=True).fit().resid
    except:
        return y.copy()

def detect_modified_zscore(r, k=3.5):
    med = np.median(r)
    mad = np.median(np.abs(r - med)) or 1e-6
    scores = 0.6745 * np.abs(r - med) / mad
    return (scores > k).astype(int), scores

def detect_rolling_stats(r, window=24, k=3.0):
    s = pd.Series(r)
    rm = s.rolling(window, min_periods=1).mean()
    rs = s.rolling(window, min_periods=1).std().fillna(1e-6)
    scores = ((s - rm) / rs).abs().values
    return (scores > k).astype(int), scores

def detect_iqr(r, factor=1.5):
    q1, q3 = np.percentile(r, 25), np.percentile(r, 75)
    iqr = q3 - q1 or 1e-6
    lo, hi = q1 - factor * iqr, q3 + factor * iqr
    dist = np.maximum(lo - r, r - hi)
    scores = np.clip(dist / iqr, 0, None)
    return (scores > 0).astype(int), scores

def detect_matrix_profile(r, m=24, discord_pct=97.5):
    if len(r) < 2 * m:
        return np.zeros(len(r)), np.zeros(len(r))

    try:
        mp = stumpy.stump(r.astype(np.float64), m=m)
    except:
        return np.zeros(len(r)), np.zeros(len(r))

    mp_vals = mp[:, 0]
    point_scores = np.zeros(len(r))

    for i, val in enumerate(mp_vals):
        point_scores[i:i+m] = np.maximum(point_scores[i:i+m], val)

    threshold = np.percentile(point_scores, discord_pct)
    return (point_scores > threshold).astype(int), point_scores

def detect_isolation_forest_features(r):
    s = pd.Series(r)
    feats = pd.DataFrame({
        "residual": r,
        "rolling_mean": s.rolling(24, min_periods=1).mean(),
        "rolling_std": s.rolling(24, min_periods=1).std().fillna(0),
        "lag_1": s.shift(1).fillna(0),
        "lag_24": s.shift(24).fillna(0),
    }).values

    X = StandardScaler().fit_transform(feats)
    clf = IsolationForest(contamination=0.05, n_estimators=150, random_state=42)

    preds = clf.fit_predict(X)
    scores = -clf.score_samples(X)

    return (preds == -1).astype(int), scores
```

```python
def evaluate_building(bid):
    path = CONSOLIDATED / f"building_{bid}_consolidated.csv"
    if not path.exists():
        print(f"Missing {path}")
        return None

    df = pd.read_csv(path, parse_dates=["timestamp"])
    df = df[df["meter_type"] == "electricity"].sort_values("timestamp")

    if df.empty:
        print(f"No electricity data for {bid}")
        return None

    y = df["meter_reading"].values
    gt = _gt(df)

    print(f"\n=== Building {bid} ===")

    r = stl_residual(y)

    methods = {
        "ModifiedZ": detect_modified_zscore,
        "RollingStats": detect_rolling_stats,
        "IQR": detect_iqr,
        "MatrixProfile": detect_matrix_profile,
        "IsolationForest": detect_isolation_forest_features,
    }

    results = []

    for name, fn in methods.items():
        t0 = time.time()
        pred, score = fn(r)
        m = metrics(pred, score, gt)
        m["method"] = name
        m["time"] = round(time.time() - t0, 2)

        print(name, "F1:", round(m["f1"], 3), "P:", round(m["precision"], 3), "R:", round(m["recall"], 3))
        results.append(m)

    return pd.DataFrame(results)
```

```python
# =========================
# INSTALL (run once)
# =========================
!pip install stumpy statsmodels scikit-learn

# =========================
# IMPORTS
# =========================
import time
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import stumpy
from sklearn.ensemble import IsolationForest
from sklearn.metrics import f1_score, precision_score, recall_score, roc_auc_score
from sklearn.preprocessing import StandardScaler
from statsmodels.tsa.seasonal import STL

warnings.filterwarnings("ignore")

# =========================
# CONFIG
# =========================
CONSOLIDATED = Path("/content/consolidated")
OUT_DIR = Path("/content/outputs/benchmark")
OUT_DIR.mkdir(parents=True, exist_ok=True)

TOP5 = [1319, 1258, 439, 247, 1225]

# =========================
# UTIL
# =========================
def _gt(df):
    return np.where(df["anomaly"].isna(), 0, df["anomaly"].fillna(0).astype(int)).astype(int)

def metrics(pred_binary, pred_scores, gt_binary):
    if gt_binary.sum() == 0:
        return dict(precision=0, recall=0, f1=0, auroc=0)

    prec = precision_score(gt_binary, pred_binary, zero_division=0)
    rec = recall_score(gt_binary, pred_binary, zero_division=0)
    f1 = f1_score(gt_binary, pred_binary, zero_division=0)

    try:
        s = (pred_scores - pred_scores.min()) / (pred_scores.max() - pred_scores.min() + 1e-6)
        auroc = roc_auc_score(gt_binary, s)
    except:
        auroc = 0

    return dict(precision=prec, recall=rec, f1=f1, auroc=auroc)

# =========================
# STEP 1: STL
# =========================
def stl_residual(y, period=24):
    if len(y) < 50:
        return y
    try:
        return STL(y, period=period, robust=True).fit().resid
    except:
        return y

# =========================
# STEP 2: FAST DETECTORS
# =========================
def detect_modified_zscore(r):
    med = np.median(r)
    mad = np.median(np.abs(r - med)) or 1e-6
    scores = 0.6745 * np.abs(r - med) / mad
    return (scores > 3.5).astype(int), scores

def detect_iqr(r):
    q1, q3 = np.percentile(r, 25), np.percentile(r, 75)
    iqr = q3 - q1 or 1e-6
    lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    scores = np.maximum(lo - r, r - hi)
    scores = np.clip(scores / iqr, 0, None)
    return (scores > 0).astype(int), scores

# =========================
# STEP 3: MATRIX PROFILE (FIXED)
# =========================
def detect_matrix_profile(y, m=8, pct=90):
    if len(y) < 2*m:
        return np.zeros(len(y)), np.zeros(len(y))

    # normalize
    y = (y - np.mean(y)) / (np.std(y) + 1e-6)

    try:
        mp = stumpy.stump(y.astype(np.float64), m=m)
    except:
        return np.zeros(len(y)), np.zeros(len(y))

    mp_vals = mp[:, 0]

    point_scores = np.zeros(len(y))
    for i, val in enumerate(mp_vals):
        point_scores[i:i+m] = np.maximum(point_scores[i:i+m], val)

    thresh = np.percentile(point_scores, pct)
    return (point_scores > thresh).astype(int), point_scores

# =========================
# STEP 4: ISOLATION FOREST
# =========================
def detect_isolation_forest(r):
    s = pd.Series(r)

    feats = pd.DataFrame({
        "r": r,
        "mean": s.rolling(24, 1).mean(),
        "std": s.rolling(24, 1).std().fillna(0),
        "lag1": s.shift(1).fillna(0),
        "lag24": s.shift(24).fillna(0)
    }).values

    X = StandardScaler().fit_transform(feats)

    clf = IsolationForest(contamination=0.05, n_estimators=150, random_state=42)
    preds = clf.fit_predict(X)
    scores = -clf.score_samples(X)

    return (preds == -1).astype(int), scores

# =========================
# EVALUATION
# =========================
def evaluate_building(bid):
    path = CONSOLIDATED / f"building_{bid}_consolidated.csv"

    if not path.exists():
        print(f"Missing {path}")
        return None

    df = pd.read_csv(path, parse_dates=["timestamp"])
    df = df[df["meter_type"] == "electricity"].sort_values("timestamp")

    if df.empty:
        print(f"No data for {bid}")
        return None

    y = df["meter_reading"].values
    gt = _gt(df)

    print(f"\n=== Building {bid} ===")

    # Step 1
    r = stl_residual(y)

    methods = {
        "ModifiedZ": lambda: detect_modified_zscore(r),
        "IQR": lambda: detect_iqr(r),
        "MatrixProfile": lambda: detect_matrix_profile(y),  # RAW SIGNAL FIX
        "IsolationForest": lambda: detect_isolation_forest(r),
    }

    rows = []

    for name, fn in methods.items():
        t0 = time.time()
        pred, score = fn()
        m = metrics(pred, score, gt)

        print(
            f"{name:<15} F1={m['f1']:.3f} "
            f"P={m['precision']:.3f} R={m['recall']:.3f}"
        )

        m["method"] = name
        m["time"] = round(time.time() - t0, 2)
        rows.append(m)

    return pd.DataFrame(rows)

# =========================
# RUN ALL
# =========================
all_results = []

for bid in TOP5:
    res = evaluate_building(bid)
    if res is not None:
        res["building_id"] = bid
        all_results.append(res)

final_df = pd.concat(all_results, ignore_index=True)

# Save
final_df.to_csv(OUT_DIR / "lead_v2_fixed.csv", index=False)

print("\nSaved results to:", OUT_DIR / "lead_v2_fixed.csv")

final_df
```

```python
# =========================
# NEW MODELS: LOF + HBOS
# =========================
from sklearn.neighbors import LocalOutlierFactor

# ---- LOF ----
def detect_lof(r, n_neighbors=20):
    r = r.reshape(-1, 1)

    try:
        lof = LocalOutlierFactor(n_neighbors=n_neighbors, contamination=0.05)
        preds = lof.fit_predict(r)
        scores = -lof.negative_outlier_factor_
        return (preds == -1).astype(int), scores
    except:
        return np.zeros(len(r)), np.zeros(len(r))


# ---- HBOS (manual implementation) ----
def detect_hbos(r, bins=30):
    try:
        hist, bin_edges = np.histogram(r, bins=bins, density=True)

        # Avoid zero densities
        hist = np.where(hist == 0, 1e-6, hist)

        # Assign each point to bin
        bin_idx = np.digitize(r, bin_edges[:-1], right=True)
        bin_idx = np.clip(bin_idx, 0, len(hist)-1)

        scores = -np.log(hist[bin_idx])

        # Normalize scores
        scores = (scores - scores.min()) / (scores.max() - scores.min() + 1e-6)

        thresh = np.percentile(scores, 95)
        return (scores > thresh).astype(int), scores

    except:
        return np.zeros(len(r)), np.zeros(len(r))


# =========================
# UPDATED EVALUATION BLOCK
# =========================
def evaluate_building_with_new_models(bid):
    path = CONSOLIDATED / f"building_{bid}_consolidated.csv"

    if not path.exists():
        print(f"Missing {path}")
        return None

    df = pd.read_csv(path, parse_dates=["timestamp"])
    df = df[df["meter_type"] == "electricity"].sort_values("timestamp")

    if df.empty:
        print(f"No data for {bid}")
        return None

    y = df["meter_reading"].values
    gt = _gt(df)

    print(f"\n=== Building {bid} (WITH NEW MODELS) ===")

    # Step 1
    r = stl_residual(y)

    methods = {
        "ModifiedZ": lambda: detect_modified_zscore(r),
        "IQR": lambda: detect_iqr(r),
        "IsolationForest": lambda: detect_isolation_forest(r),
        "LOF": lambda: detect_lof(r),                 # NEW
        "HBOS": lambda: detect_hbos(r),               # NEW
    }

    results = {}
    rows = []

    # Run all models
    for name, fn in methods.items():
        t0 = time.time()
        pred, score = fn()
        m = metrics(pred, score, gt)

        print(
            f"{name:<15} F1={m['f1']:.3f} "
            f"P={m['precision']:.3f} R={m['recall']:.3f}"
        )

        results[name] = pred

        m["method"] = name
        m["time"] = round(time.time() - t0, 2)
        rows.append(m)

    # =========================
    # ENSEMBLE (IMPORTANT)
    # =========================
    # Rule:
    # (IsolationForest OR LOF) AND (Z-score OR IQR)

    iso = results["IsolationForest"]
    lof = results["LOF"]
    z   = results["ModifiedZ"]
    iqr = results["IQR"]

    ensemble_pred = ((iso | lof) & (z | iqr)).astype(int)

    # simple score = vote fraction
    vote_sum = iso + lof + z + iqr
    ensemble_score = vote_sum / 4.0

    m = metrics(ensemble_pred, ensemble_score, gt)

    print("\n🔥 Ensemble")
    print(
        f"{'Ensemble':<15} F1={m['f1']:.3f} "
        f"P={m['precision']:.3f} R={m['recall']:.3f}"
    )

    m["method"] = "Ensemble"
    rows.append(m)

    return pd.DataFrame(rows)
```

```python
# =========================
# NEW MODELS: KDE + GMM + ECOD
# =========================
from sklearn.neighbors import KernelDensity
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import QuantileTransformer

# ---- KDE ----
def detect_kde(r, bandwidth=0.5):
    r = r.reshape(-1, 1)

    try:
        kde = KernelDensity(kernel='gaussian', bandwidth=bandwidth)
        kde.fit(r)

        log_density = kde.score_samples(r)
        scores = -log_density  # lower density = more anomalous

        # normalize
        scores = (scores - scores.min()) / (scores.max() - scores.min() + 1e-6)

        thresh = np.percentile(scores, 95)
        return (scores > thresh).astype(int), scores

    except:
        return np.zeros(len(r)), np.zeros(len(r))


# ---- GMM ----
def detect_gmm(r, n_components=3):
    r = r.reshape(-1, 1)

    try:
        gmm = GaussianMixture(n_components=n_components, covariance_type='full', random_state=42)
        gmm.fit(r)

        log_prob = gmm.score_samples(r)
        scores = -log_prob  # lower likelihood = anomaly

        scores = (scores - scores.min()) / (scores.max() - scores.min() + 1e-6)

        thresh = np.percentile(scores, 95)
        return (scores > thresh).astype(int), scores

    except:
        return np.zeros(len(r)), np.zeros(len(r))


# ---- ECOD-like (fast approximation) ----
def detect_ecod(r):
    try:
        # rank-based transformation (copula-like)
        qt = QuantileTransformer(output_distribution='uniform')
        r_trans = qt.fit_transform(r.reshape(-1, 1)).flatten()

        # tail probability score
        scores = np.minimum(r_trans, 1 - r_trans)

        # invert so rare = high score
        scores = 1 - scores

        scores = (scores - scores.min()) / (scores.max() - scores.min() + 1e-6)

        thresh = np.percentile(scores, 95)
        return (scores > thresh).astype(int), scores

    except:
        return np.zeros(len(r)), np.zeros(len(r))


# =========================
# UPDATED EVALUATION WITH NEW MODELS
# =========================
def evaluate_building_advanced(bid):
    path = CONSOLIDATED / f"building_{bid}_consolidated.csv"

    if not path.exists():
        print(f"Missing {path}")
        return None

    df = pd.read_csv(path, parse_dates=["timestamp"])
    df = df[df["meter_type"] == "electricity"].sort_values("timestamp")

    if df.empty:
        print(f"No data for {bid}")
        return None

    y = df["meter_reading"].values
    gt = _gt(df)

    print(f"\n=== Building {bid} (ADVANCED MODELS) ===")

    # residual for classical models
    r = stl_residual(y)

    methods = {
        "ModifiedZ": lambda: detect_modified_zscore(r),
        "IQR": lambda: detect_iqr(r),
        "IsolationForest": lambda: detect_isolation_forest(r),
        "HBOS": lambda: detect_hbos(r),

        # NEW
        "KDE": lambda: detect_kde(r),
        "GMM": lambda: detect_gmm(r),
        "ECOD": lambda: detect_ecod(r),
    }

    preds_dict = {}
    scores_dict = {}
    rows = []

    # Run models
    for name, fn in methods.items():
        t0 = time.time()
        pred, score = fn()
        m = metrics(pred, score, gt)

        print(f"{name:<15} F1={m['f1']:.3f} P={m['precision']:.3f} R={m['recall']:.3f}")

        preds_dict[name] = pred
        scores_dict[name] = score

        m["method"] = name
        m["time"] = round(time.time() - t0, 2)
        rows.append(m)

    # =========================
    # SMART ENSEMBLE (score fusion)
    # =========================
    def ensemble_score_fusion(scores_dict):
        norm_scores = {}
        for k, s in scores_dict.items():
            s = (s - s.min()) / (s.max() - s.min() + 1e-6)
            norm_scores[k] = s

        # tuned weights (based on your observations)
        weights = {
            "HBOS": 0.25,
            "KDE": 0.20,
            "GMM": 0.20,
            "IsolationForest": 0.15,
            "ECOD": 0.10,
            "ModifiedZ": 0.05,
            "IQR": 0.05,
        }

        final_score = sum(weights[k] * norm_scores[k] for k in weights if k in norm_scores)

        thresh = np.percentile(final_score, 92)
        pred = (final_score > thresh).astype(int)

        return pred, final_score

    ensemble_pred, ensemble_score = ensemble_score_fusion(scores_dict)
    m = metrics(ensemble_pred, ensemble_score, gt)

    print("\n🔥 Ensemble (Advanced)")
    print(f"{'Ensemble':<15} F1={m['f1']:.3f} P={m['precision']:.3f} R={m['recall']:.3f}")

    m["method"] = "Ensemble_Advanced"
    rows.append(m)

    return pd.DataFrame(rows)
```

```python
# =========================
# FEATURE ENGINEERING + MODELS
# =========================

def build_features(y, r):
    s = pd.Series(r)

    # --- core ---
    feat = pd.DataFrame({
        "residual": r,
    })

    # --- rolling stats ---
    feat["roll_mean_24"] = s.rolling(24, 1).mean()
    feat["roll_std_24"]  = s.rolling(24, 1).std().fillna(0)

    # deviation from local mean
    feat["dev_mean"] = r - feat["roll_mean_24"]

    # --- temporal change (VERY important) ---
    feat["diff_1"] = s.diff().fillna(0)
    feat["abs_diff_1"] = feat["diff_1"].abs()

    # --- seasonal deviation ---
    feat["diff_24"] = s.diff(24).fillna(0)

    # --- normalized rank (distribution info) ---
    feat["rank"] = s.rank(pct=True)

    return feat.fillna(0).values


# =========================
# MODELS ON FEATURES
# =========================
def detect_iforest_features(X):
    try:
        Xs = StandardScaler().fit_transform(X)
        clf = IsolationForest(contamination=0.05, n_estimators=150, random_state=42)
        pred = clf.fit_predict(Xs)
        scores = -clf.score_samples(Xs)
        return (pred == -1).astype(int), scores
    except:
        return np.zeros(len(X)), np.zeros(len(X))


def detect_gmm_features(X, n_components=3):
    try:
        Xs = StandardScaler().fit_transform(X)
        gmm = GaussianMixture(n_components=n_components, random_state=42)
        gmm.fit(Xs)

        log_prob = gmm.score_samples(Xs)
        scores = -log_prob

        scores = (scores - scores.min()) / (scores.max() - scores.min() + 1e-6)
        thresh = np.percentile(scores, 95)

        return (scores > thresh).astype(int), scores
    except:
        return np.zeros(len(X)), np.zeros(len(X))


# =========================
# EVALUATION (FEATURE VERSION)
# =========================
def evaluate_building_features(bid):
    path = CONSOLIDATED / f"building_{bid}_consolidated.csv"

    if not path.exists():
        print(f"Missing {path}")
        return None

    df = pd.read_csv(path, parse_dates=["timestamp"])
    df = df[df["meter_type"] == "electricity"].sort_values("timestamp")

    if df.empty:
        print(f"No data for {bid}")
        return None

    y = df["meter_reading"].values
    gt = _gt(df)

    print(f"\n=== Building {bid} (FEATURE MODELS) ===")

    # residual
    r = stl_residual(y)

    # 🔥 FEATURE MATRIX
    X = build_features(y, r)

    methods = {
        "IsolationForest_feat": lambda: detect_iforest_features(X),
        "GMM_feat": lambda: detect_gmm_features(X),
    }

    scores_dict = {}
    rows = []

    for name, fn in methods.items():
        t0 = time.time()
        pred, score = fn()
        m = metrics(pred, score, gt)

        print(f"{name:<20} F1={m['f1']:.3f} P={m['precision']:.3f} R={m['recall']:.3f}")

        scores_dict[name] = score

        m["method"] = name
        m["time"] = round(time.time() - t0, 2)
        rows.append(m)

    # =========================
    # SIMPLE BEST-OF ENSEMBLE
    # =========================
    # take max score across models (very strong baseline)
    stacked_scores = np.vstack(list(scores_dict.values()))
    ensemble_score = stacked_scores.max(axis=0)

    thresh = np.percentile(ensemble_score, 92)
    ensemble_pred = (ensemble_score > thresh).astype(int)

    m = metrics(ensemble_pred, ensemble_score, gt)

    print("\n🔥 Ensemble (Feature Max Fusion)")
    print(f"{'Ensemble':<20} F1={m['f1']:.3f} P={m['precision']:.3f} R={m['recall']:.3f}")

    m["method"] = "Ensemble_Feature"
    rows.append(m)

    return pd.DataFrame(rows)
```

```python
# =========================
# WINDOW-BASED FEATURES + IF
# =========================

def build_window_features(r, window=12):
    """
    Build sliding window features.
    Each point gets context of surrounding values.
    """
    n = len(r)
    pad = window // 2

    # pad edges
    r_pad = np.pad(r, (pad, pad), mode='edge')

    X = []
    for i in range(n):
        w = r_pad[i:i+window]

        # raw window
        feats = list(w)

        # summary stats (VERY important)
        feats += [
            np.mean(w),
            np.std(w),
            np.max(w) - np.min(w),
            w[-1] - w[0],  # slope
        ]

        X.append(feats)

    return np.array(X)


def detect_iforest_windows(r, window=12):
    try:
        X = build_window_features(r, window)

        Xs = StandardScaler().fit_transform(X)

        clf = IsolationForest(
            contamination=0.05,
            n_estimators=150,
            random_state=42
        )

        pred = clf.fit_predict(Xs)
        scores = -clf.score_samples(Xs)

        return (pred == -1).astype(int), scores

    except:
        return np.zeros(len(r)), np.zeros(len(r))


# =========================
# TEMPORAL SMOOTHING
# =========================
def smooth_predictions(pred, min_run=3):
    """
    Remove isolated spikes, keep sustained anomalies
    """
    pred = pred.copy()
    n = len(pred)

    i = 0
    while i < n:
        if pred[i] == 1:
            j = i
            while j < n and pred[j] == 1:
                j += 1

            if (j - i) < min_run:
                pred[i:j] = 0

            i = j
        else:
            i += 1

    return pred


# =========================
# EVALUATION
# =========================
def evaluate_building_window(bid):
    path = CONSOLIDATED / f"building_{bid}_consolidated.csv"

    if not path.exists():
        print(f"Missing {path}")
        return None

    df = pd.read_csv(path, parse_dates=["timestamp"])
    df = df[df["meter_type"] == "electricity"].sort_values("timestamp")

    if df.empty:
        print(f"No data for {bid}")
        return None

    y = df["meter_reading"].values
    gt = _gt(df)

    print(f"\n=== Building {bid} (WINDOW MODEL) ===")

    # STL residual
    r = stl_residual(y)

    # Window-based IF
    t0 = time.time()
    pred, score = detect_iforest_windows(r, window=12)

    # smoothing (IMPORTANT)
    pred_smooth = smooth_predictions(pred, min_run=3)

    m = metrics(pred_smooth, score, gt)

    print(
        f"{'Window_IF':<15} F1={m['f1']:.3f} "
        f"P={m['precision']:.3f} R={m['recall']:.3f}"
    )

    return pd.DataFrame([{
        "method": "Window_IF",
        **m,
        "time": round(time.time() - t0, 2)
    }])
```

```python
# =========================
# HYBRID MODEL: WINDOW + HBOS
# =========================

from pyod.models.hbos import HBOS

# --- window features (same as before, slightly enhanced) ---
def build_window_features_hybrid(r, window=12):
    n = len(r)
    pad = window // 2

    r_pad = np.pad(r, (pad, pad), mode='edge')
    X = []

    global_median = np.median(r)

    for i in range(n):
        w = r_pad[i:i+window]

        feats = list(w)

        # --- temporal structure ---
        feats += [
            np.mean(w),
            np.std(w),
            np.max(w) - np.min(w),
            w[-1] - w[0],  # slope
        ]

        # --- density / distribution ---
        feats += [
            np.mean(np.abs(w)),
            np.percentile(w, 90),
            np.percentile(w, 10),
            w[-1] - global_median,  # global deviation
        ]

        X.append(feats)

    return np.array(X)


# --- window IF ---
def detect_window_if(r, window=12):
    try:
        X = build_window_features_hybrid(r, window)
        Xs = StandardScaler().fit_transform(X)

        clf = IsolationForest(
            contamination=0.05,
            n_estimators=150,
            random_state=42
        )

        pred = clf.fit_predict(Xs)
        score = -clf.score_samples(Xs)

        return (pred == -1).astype(int), score

    except:
        return np.zeros(len(r)), np.zeros(len(r))


# --- HBOS ---
def detect_hbos(r):
    try:
        X = r.reshape(-1, 1)
        clf = HBOS(contamination=0.05)
        clf.fit(X)

        score = clf.decision_scores_
        pred = clf.predict(X)

        return pred.astype(int), score

    except:
        return np.zeros(len(r)), np.zeros(len(r))


# --- smoothing ---
def smooth_predictions(pred, min_run=3):
    pred = pred.copy()
    n = len(pred)

    i = 0
    while i < n:
        if pred[i] == 1:
            j = i
            while j < n and pred[j] == 1:
                j += 1

            if (j - i) < min_run:
                pred[i:j] = 0

            i = j
        else:
            i += 1

    return pred


# =========================
# EVALUATION
# =========================
def evaluate_building_hybrid(bid):
    path = CONSOLIDATED / f"building_{bid}_consolidated.csv"

    if not path.exists():
        print(f"Missing {path}")
        return None

    df = pd.read_csv(path, parse_dates=["timestamp"])
    df = df[df["meter_type"] == "electricity"].sort_values("timestamp")

    if df.empty:
        print(f"No data for {bid}")
        return None

    y = df["meter_reading"].values
    gt = _gt(df)

    print(f"\n=== Building {bid} (HYBRID MODEL) ===")

    # STL residual
    r = stl_residual(y)

    # --- models ---
    pred_w, score_w = detect_window_if(r)
    pred_h, score_h = detect_hbos(r)

    # normalize scores
    def norm(s):
        return (s - s.min()) / (s.max() - s.min() + 1e-6)

    sw = norm(score_w)
    sh = norm(score_h)

    # 🔥 weighted fusion
    final_score = 0.6 * sw + 0.4 * sh

    thresh = np.percentile(final_score, 92)
    pred = (final_score > thresh).astype(int)

    # smoothing
    pred = smooth_predictions(pred, min_run=3)

    m = metrics(pred, final_score, gt)

    print(
        f"{'Hybrid':<15} F1={m['f1']:.3f} "
        f"P={m['precision']:.3f} R={m['recall']:.3f}"
    )

    return pd.DataFrame([{
        "method": "Hybrid_Window_HBOS",
        **m
    }])
```

```python
all_results = []

for bid in TOP5:
    res = evaluate_building_hybrid(bid)
    if res is not None:
        res["building_id"] = bid
        all_results.append(res)

final_df = pd.concat(all_results, ignore_index=True)
final_df
```

```python
# =========================
# PURE NUMPY SPECTRAL RESIDUAL
# =========================

import numpy as np
import pandas as pd


def spectral_residual_scores(y, eps=1e-8):
    """
    Compute Spectral Residual saliency scores.
    Based on:
    Ren et al. (Microsoft) - Time-Series Anomaly Detection Service
    """
    y = y.astype(float)

    # FFT
    fft = np.fft.fft(y)
    mag = np.abs(fft)
    log_mag = np.log(mag + eps)

    # moving average of log spectrum
    avg_log_mag = pd.Series(log_mag).rolling(window=3, center=True, min_periods=1).mean().values

    # spectral residual
    spectral_residual = log_mag - avg_log_mag

    # reconstruct signal
    fft_sr = np.exp(spectral_residual) * np.exp(1j * np.angle(fft))
    saliency_map = np.abs(np.fft.ifft(fft_sr))

    return saliency_map


def detect_spectral_residual(y, threshold_pct=95):
    scores = spectral_residual_scores(y)

    # normalize
    scores = (scores - scores.min()) / (scores.max() - scores.min() + 1e-8)

    threshold = np.percentile(scores, threshold_pct)
    pred = (scores > threshold).astype(int)

    return pred, scores


# =========================
# OPTIONAL SMOOTHING
# =========================
def smooth_predictions(pred, min_run=3):
    pred = pred.copy()
    n = len(pred)

    i = 0
    while i < n:
        if pred[i] == 1:
            j = i
            while j < n and pred[j] == 1:
                j += 1

            if (j - i) < min_run:
                pred[i:j] = 0

            i = j
        else:
            i += 1

    return pred


# =========================
# EVALUATION
# =========================
def evaluate_building_sr(bid):
    path = CONSOLIDATED / f"building_{bid}_consolidated.csv"

    if not path.exists():
        print(f"Missing {path}")
        return None

    df = pd.read_csv(path, parse_dates=["timestamp"])
    df = df[df["meter_type"] == "electricity"].sort_values("timestamp")

    if df.empty:
        print(f"No data for {bid}")
        return None

    y = df["meter_reading"].values
    gt = _gt(df)

    print(f"\n=== Building {bid} (Spectral Residual - NumPy) ===")

    pred, score = detect_spectral_residual(y, threshold_pct=95)

    # smoothing helps reduce noise
    pred = smooth_predictions(pred, min_run=3)

    m = metrics(pred, score, gt)

    print(
        f"{'SpectralResidual':<20} "
        f"F1={m['f1']:.3f} P={m['precision']:.3f} R={m['recall']:.3f}"
    )

    return pd.DataFrame([{
        "method": "SpectralResidual",
        **m
    }])
```
